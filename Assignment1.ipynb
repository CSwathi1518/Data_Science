```python
n=int(input())
if n > 1:
    for i in range(2,int(n**0.5)+1):
        if n % i == 0:
            print("Not Prime Number")
            break
    else:
        print("Prime Number")
else:
    print("Not prime")
```

```python
import random
a = random.randint(1, 10)
b = random.randint(1, 10)
print("What is the product of", a, "and", b, "?")
ans = int(input("Enter your answer: "))
if ans == a * b:
    print("Correct!")
else:
    print("Wrong! The answer is", a * b)
```

```python
choice = input("Do you want odd or even numbers? (Enter 'odd' or 'even'): ").lower()

if choice == 'even':
    for i in range(100, 201):
        if i % 2 == 0:
            print("Square of", i, "is", i**2)
elif choice == 'odd':
    for i in range(100, 201):
        if i % 2 != 0:  # checks if the number is odd
            print("Square of", i, "is", i**2)
else:
    print("Invalid input! Please enter 'odd' or 'even'.")
```

```python
room_charge_per_day = 2000
doctor_fee = 1500
lab_test_charge = 300

# Patient details (given in question)
days_stayed = 4
num_tests = 3
medicine_charges = 2400

# Calculate total before discount
total_before_discount = (days_stayed * room_charge_per_day) + doctor_fee + (num_tests * lab_test_charge) + medicine_charges

# Apply discount if total > 10,000
if total_before_discount > 10000:
    discount = total_before_discount * 0.10
else:
    discount = 0

# Final bill after discount
final_bill = total_before_discount - discount

# Display the bill
print("----- HOSPITAL BILL -----")
print("Room Charges (₹2000/day x", days_stayed, "): ₹", days_stayed * room_charge_per_day)
print("Doctor Consultation Fee: ₹", doctor_fee)
print("Lab Test Charges (₹300 x", num_tests, "): ₹", num_tests * lab_test_charge)
print("Medicine Charges: ₹", medicine_charges)
print("--------------------------")
print("Total Before Discount: ₹", total_before_discount)
print("Discount Applied: ₹", discount)
print("--------------------------")
print("Final Bill: ₹", final_bill)
print("--------------------------")
```
