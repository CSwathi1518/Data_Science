```python
# ========================================
# STEP 1: Load and show summary
# ========================================
print("STEP 1: Load the patient dataset and show summary\n")
patient_df = pd.read_csv(url)
print("Patient Dataset:")
print(patient_df)
print("\nDataset Information:")
patient_df.info()
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 2: Select billing columns
# ========================================
print("STEP 2: Select only billing-relevant columns\n")
billing_columns = ['PatientID', 'Department', 'Doctor', 'BillAmount']
patient_billing = patient_df[billing_columns].copy()
print(patient_billing)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 3: Drop administrative columns
# ========================================
print("STEP 3: Drop administrative columns\n")
patient_df_clean = patient_df.drop(['ReceptionistID', 'CheckInTime'], axis=1)
print(patient_df_clean)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 4: Group by department
# ========================================
print("STEP 4: Total bill amount per department\n")
department_revenue = patient_df_clean.groupby('Department')['BillAmount'].sum()
print("Total Bill Amount per Department:")
print(department_revenue)
print("\nDetailed Department Summary:")
dept_summary = patient_df_clean.groupby('Department').agg({
    'BillAmount': ['sum', 'count', 'mean']
})
print(dept_summary)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 5: Remove duplicates
# ========================================
print("STEP 5: Remove duplicate patient records\n")
patient_df_unique = patient_df_clean.drop_duplicates(subset='PatientID', keep='first')
print(patient_df_unique)
print(f"\nRows before: {len(patient_df_clean)}, Rows after: {len(patient_df_unique)}")
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 6: Fill missing values with mean
# ========================================
print("STEP 6: Fill missing BillAmount with mean\n")
mean_bill = patient_df_unique['BillAmount'].mean()
print(f"Mean Bill Amount: ${mean_bill:.2f}\n")

patient_df_filled = patient_df_unique.copy()
patient_df_filled['BillAmount'] = patient_df_filled['BillAmount'].fillna(mean_bill)
print("After Filling Missing Values:")
print(patient_df_filled)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 7: Merge billing dataset
# ========================================
print("STEP 7: Merge billing dataset with patient dataset\n")
billing_df = pd.read_csv(url)
print("Billing Dataset:")
print(billing_df)
print("\n")

merged_df = pd.merge(patient_df_filled, billing_df, on='PatientID', how='left')
print("Merged Dataset:")
print(merged_df)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 8: Concatenate new patients (row-wise)
# ========================================
print("STEP 8: Add new patients for current week (row-wise concatenation)\n")
new_patients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['Frank', 'Grace'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. Smith', 'Dr. John'],
    'BillAmount': [5500.0, 4800.0],
    'InsuranceCovered': [2200, 2000],
    'FinalAmount': [3300, 2800]
})

print("New Patients:")
print(new_patients)
print("\n")

final_df = pd.concat([merged_df, new_patients], ignore_index=True)
print("After Adding New Patients:")
print(final_df)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# STEP 9: Add billing category columns (column-wise)
# ========================================
print("STEP 9: Add new billing category columns (column-wise concatenation)\n")
additional_billing = pd.DataFrame({
    'PaymentStatus': ['Paid', 'Pending', 'Paid', 'Paid', 'Pending', 'Paid', 'Paid'],
    'PaymentMethod': ['Card', 'Cash', 'Insurance', 'Card', 'Cash', 'Card', 'Insurance']
})

final_complete_df = pd.concat([final_df, additional_billing], axis=1)
print("Final Complete Dataset:")
print(final_complete_df)
print("\n" + "="*70 + "\n")
```

```python
# ========================================
# FINAL ANALYSIS
# ========================================
print("="*70)
print("FINAL ANALYSIS SUMMARY")
print("="*70 + "\n")

print("1. Department-wise Total Revenue:")
dept_revenue = final_complete_df.groupby('Department')['BillAmount'].sum().sort_values(ascending=False)
print(dept_revenue)
print("\n")

print("2. Doctor-wise Performance:")
doctor_perf = final_complete_df.groupby('Doctor').agg({
    'PatientID': 'count',
    'BillAmount': 'sum',
    'FinalAmount': 'sum'
}).rename(columns={'PatientID': 'Patient_Count'})
print(doctor_perf)
print("\n")

print("3. Overall Statistics:")
print(f"   Total Patients: {len(final_complete_df)}")
print(f"   Total Revenue (Bill Amount): ${final_complete_df['BillAmount'].sum():,.2f}")
print(f"   Total After Insurance: ${final_complete_df['FinalAmount'].sum():,.2f}")
print(f"   Average Bill per Patient: ${final_complete_df['BillAmount'].mean():,.2f}")
print(f"   Total Insurance Coverage: ${final_complete_df['InsuranceCovered'].sum():,.2f}")
print("\n")

print("4. Payment Status Summary:")
print(final_complete_df['PaymentStatus'].value_counts())
print("\n")

# Save final dataset
final_complete_df.to_csv('Final_Hospital_Analysis.csv', index=False)
print("✓ Final dataset saved as 'Final_Hospital_Analysis.csv'")
```
