Problem 1: Hypothesis Testing (Bombay Hospitality Ltd.)

Key Steps:

Hypotheses Setup:

H₀: μ = $4,000 (costs follow the model)

H₁: μ > $4,000 (costs are higher - this is what restaurant owners claim)

This is a right-tailed test since we're testing if costs are higher

Test Statistic Calculation:

Theoretical mean: W = 1,000 + 5(600) = $4,000

Since W = 1,000 + 5X, the standard deviation of W = 5 × 25 = 125

Z-statistic measures how many standard errors the sample mean is from the theoretical mean

Decision Criteria:

Compare z-statistic with critical value (1.645 for α = 0.05)
Also compare p-value with α

If z > critical value OR p-value < α, reject H₀

The Result: The test statistic z = -38, which is far in the negative direction, indicating the observed costs ($3,050) are actually lower than predicted ($4,000), not higher!

```python
import numpy as np
from scipy import stats
import math
```

```python
# Given data
sample_mean = 3050  # Rs.
sample_size = 25
X_mean = 600  # units
X_std = 25  # units
alpha = 0.05

print("\n GIVEN INFORMATION:")
print(f"   • Cost Model: W = $1,000 + $5X")
print(f"   • Sample mean weekly cost: Rs. {sample_mean}")
print(f"   • Sample size: {sample_size} restaurants")
print(f"   • Mean units produced (X): {X_mean} units")
print(f"   • Standard deviation of X: {X_std} units")
print(f"   • Significance level (α): {alpha}")
```

```python
# Step 1: State the Hypotheses
print("\n" + "="*70)
print("STEP 1: STATE THE HYPOTHESES")
print("="*70)

# Calculate theoretical mean
theoretical_mean = 1000 + 5 * X_mean
print(f"\nTheoretical mean cost = $1,000 + $5(600) = ${theoretical_mean}")

print("\n Null Hypothesis (H₀): μ = $4,000")
print("   The weekly operating costs follow the theoretical model.")
print("\n Alternative Hypothesis (H₁): μ > $4,000")
print("   The weekly operating costs are HIGHER than the model suggests.")
print("\n This is a RIGHT-TAILED TEST (one-tailed)")
```

```python
# Step 2: Calculate the Test Statistic
print("\n" + "="*70)
print("STEP 2: CALCULATE THE TEST STATISTIC")
print("="*70)

# Calculate standard deviation of W
# Since W = 1000 + 5X, and X ~ N(600, 25²)
# Standard deviation of W = 5 * standard deviation of X
sigma_w = 5 * X_std
print(f"\nStandard deviation of W (σ_W) = 5 × {X_std} = {sigma_w}")

# Standard error
standard_error = sigma_w / math.sqrt(sample_size)
print(f"Standard error = σ_W / √n = {sigma_w} / √{sample_size} = {standard_error}")

# Calculate z-statistic
z_statistic = (sample_mean - theoretical_mean) / standard_error

print(f"\n Z-Statistic Formula:")
print(f"   z = (x̄ - μ) / (σ / √n)")
print(f"   z = ({sample_mean} - {theoretical_mean}) / {standard_error}")
print(f"   z = {sample_mean - theoretical_mean} / {standard_error}")
print(f"   z = {z_statistic:.4f}")
```

```python
# Step 3: Determine the Critical Value and P-value
print("\n" + "="*70)
print("STEP 3: DETERMINE CRITICAL VALUE AND P-VALUE")
print("="*70)

# Critical value for right-tailed test at α = 0.05
critical_value = stats.norm.ppf(1 - alpha)
print(f"\n For α = {alpha} (right-tailed test):")
print(f"   Critical value (z_critical) = {critical_value:.4f}")

# Calculate p-value
p_value = 1 - stats.norm.cdf(z_statistic)
print(f"\n P-value = P(Z > {z_statistic:.4f}) = {p_value:.6f}")
```

```python
# Step 4: Make a Decision
print("\n" + "="*70)
print("STEP 4: MAKE A DECISION")
print("="*70)

print(f"\n Decision Criteria:")
print(f"   • Test statistic (z) = {z_statistic:.4f}")
print(f"   • Critical value = {critical_value:.4f}")
print(f"   • P-value = {p_value:.6f}")
print(f"   • Significance level (α) = {alpha}")

print(f"\n Decision Rules:")
print(f"   1. If z > z_critical, REJECT H₀")
print(f"   2. If p-value < α, REJECT H₀")

if z_statistic > critical_value:
    decision = "REJECT"
    print(f"\n Since {z_statistic:.4f} > {critical_value:.4f}, we REJECT H₀")
else:
    decision = "FAIL TO REJECT"
    print(f"\n Since {z_statistic:.4f} ≤ {critical_value:.4f}, we FAIL TO REJECT H₀")

if p_value < alpha:
    print(f" Since p-value ({p_value:.6f}) < α ({alpha}), we REJECT H₀")
else:
    print(f" Since p-value ({p_value:.6f}) ≥ α ({alpha}), we FAIL TO REJECT H₀")
```

```python
# Step 5: Conclusion
print("\n" + "="*70)
print("STEP 5: CONCLUSION")
print("="*70)

if decision == "REJECT":
    print(f"\n CONCLUSION:")
    print(f"   There IS strong statistical evidence at the {alpha} significance level")
    print(f"   to support the restaurant owners' claim that the weekly operating")
    print(f"   costs are HIGHER than what the theoretical model suggests.")
    print(f"\n   The observed mean cost of Rs. {sample_mean} is significantly greater")
    print(f"   than the theoretical mean of Rs. {theoretical_mean}.")
else:
    print(f"\n CONCLUSION:")
    print(f"   There is NOT sufficient statistical evidence at the {alpha} significance")
    print(f"   level to support the restaurant owners' claim. The observed costs")
    print(f"   could be due to random variation.")
```

Problem 2: Confidence Intervals (Print-Head Durability)

Part A - Using Sample Standard Deviation (t-distribution):

Why t-distribution?

Small sample (n=15 < 30)

Population σ is unknown

More uncertainty → wider intervals

Steps:

Calculate sample mean and sample standard deviation

Find t-critical value with df = 14

Calculate margin of error: t × (s/√n)

CI = mean ± margin of error

Part B - Using Known Population Standard Deviation (z-distribution):

Why z-distribution?

Population σ is known (given as 0.2)

Use standard normal distribution

Narrower interval due to less uncertainty

Key Insight: The z-interval is narrower because knowing the true population standard deviation removes estimation uncertainty.

```python
# ============================================================================
# PROBLEM 2: CONFIDENCE INTERVALS FOR PRINT-HEAD DURABILITY
# ============================================================================

print("\n\n" + "="*70)
print("PROBLEM 2: CONFIDENCE INTERVALS - PRINT-HEAD DURABILITY")
print("="*70)

# Given data
durability_data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32,
                   0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
n = len(durability_data)
confidence_level = 0.99
alpha_ci = 1 - confidence_level

print("\n GIVEN DATA:")
print(f"   Sample size: {n} print-heads")
print(f"   Confidence level: {confidence_level*100}%")
print(f"   Durability data (millions of characters):")
print(f"   {durability_data}")
```

```python
# Calculate sample statistics
sample_mean_durability = np.mean(durability_data)
sample_std = np.std(durability_data, ddof=1)  # ddof=1 for sample std

print(f"\n Sample Statistics:")
print(f"   Sample mean (x̄) = {sample_mean_durability:.4f} million characters")
print(f"   Sample standard deviation (s) = {sample_std:.4f} million characters")
```

```python
# Part A: Using Sample Standard Deviation (t-distribution)
print("\n" + "="*70)
print("PART A: 99% CI USING SAMPLE STANDARD DEVIATION (t-distribution)")
print("="*70)

print("\n Why use t-distribution?")
print("   • Small sample size (n = 15 < 30)")
print("   • Population standard deviation is UNKNOWN")
print("   • We use sample standard deviation as an estimate")
print("   • This adds uncertainty, so we use t-distribution")
```

```python
# Degrees of freedom
df = n - 1
print(f"\n Degrees of freedom = n - 1 = {n} - 1 = {df}")

# t-critical value
t_critical = stats.t.ppf(1 - alpha_ci/2, df)
print(f"   t-critical value (two-tailed, α = {alpha_ci}) = {t_critical:.4f}")

# Standard error
se_t = sample_std / math.sqrt(n)
print(f"\n Standard error = s / √n = {sample_std:.4f} / √{n} = {se_t:.4f}")

# Margin of error
margin_error_t = t_critical * se_t
print(f"   Margin of error = t × SE = {t_critical:.4f} × {se_t:.4f} = {margin_error_t:.4f}")

# Confidence interval
ci_lower_t = sample_mean_durability - margin_error_t
ci_upper_t = sample_mean_durability + margin_error_t

print(f"\n 99% CONFIDENCE INTERVAL (using t-distribution):")
print(f"   CI = x̄ ± (t × s/√n)")
print(f"   CI = {sample_mean_durability:.4f} ± {margin_error_t:.4f}")
print(f"   CI = [{ci_lower_t:.4f}, {ci_upper_t:.4f}] million characters")
print(f"\n INTERPRETATION:")
print(f"   We are 99% confident that the true mean durability of print-heads")
print(f"   lies between {ci_lower_t:.4f} and {ci_upper_t:.4f} million characters.")

# Part B: Using Known Population Standard Deviation (z-distribution)
print("\n" + "="*70)
print("PART B: 99% CI USING KNOWN POPULATION STANDARD DEVIATION (z-distribution)")
print("="*70)

population_std = 0.2
print(f"\n Given: Population standard deviation (σ) = {population_std} million characters")

print("\n Why use z-distribution?")
print("   • Population standard deviation is KNOWN")
print("   • Sample size doesn't matter when σ is known")
print("   • Use standard normal (z) distribution")

# z-critical value
z_critical_ci = stats.norm.ppf(1 - alpha_ci/2)
print(f"\n z-critical value (two-tailed, α = {alpha_ci}) = {z_critical_ci:.4f}")

# Standard error with known population std
se_z = population_std / math.sqrt(n)
print(f"   Standard error = σ / √n = {population_std} / √{n} = {se_z:.4f}")

# Margin of error
margin_error_z = z_critical_ci * se_z
print(f"   Margin of error = z × SE = {z_critical_ci:.4f} × {se_z:.4f} = {margin_error_z:.4f}")

# Confidence interval
ci_lower_z = sample_mean_durability - margin_error_z
ci_upper_z = sample_mean_durability + margin_error_z

print(f"\n 99% CONFIDENCE INTERVAL (using z-distribution):")
print(f"   CI = x̄ ± (z × σ/√n)")
print(f"   CI = {sample_mean_durability:.4f} ± {margin_error_z:.4f}")
print(f"   CI = [{ci_lower_z:.4f}, {ci_upper_z:.4f}] million characters")
print(f"\n INTERPRETATION:")
print(f"   We are 99% confident that the true mean durability of print-heads")
print(f"   lies between {ci_lower_z:.4f} and {ci_upper_z:.4f} million characters.")
```

```python
# Comparison
print("\n" + "="*70)
print("COMPARISON OF BOTH CONFIDENCE INTERVALS")
print("="*70)
print(f"\nUsing t-distribution (unknown σ): [{ci_lower_t:.4f}, {ci_upper_t:.4f}]")
print(f"Using z-distribution (known σ):   [{ci_lower_z:.4f}, {ci_upper_z:.4f}]")
print(f"\nWidth of t-interval: {ci_upper_t - ci_lower_t:.4f}")
print(f"Width of z-interval: {ci_upper_z - ci_lower_z:.4f}")
print(f"\n OBSERVATION:")
if (ci_upper_t - ci_lower_t) > (ci_upper_z - ci_lower_z):
    print(f"   The t-interval is WIDER than the z-interval.")
    print(f"   This reflects the additional uncertainty when estimating σ from the sample.")
else:
    print(f"   The z-interval is wider, which is unusual but can happen")
    print(f"   if the sample std is smaller than the population std.")

print("\n" + "="*70)
print("ANALYSIS COMPLETE!")
print("="*70)
```
