#Section C: Pandas

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve',
             'Frank', 'Grace', 'Henry', 'Ivy', 'Jack'],
    'Age': [25, 30, 35, 28, 32, 29, 27, 31, 26, 33],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix',
             'Philadelphia', 'San Antonio', 'San Diego', 'Dallas', 'New York'],
    'Salary': [50000, 60000, np.nan, 55000, 62000,
               58000, np.nan, 65000, 52000, 70000]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
cl=df[["Name","Age"]]
cl
```

```python
dr_drop=df.drop("City",axis=1)
dr_drop
```

```python
salary_mean = dr_drop['Salary'].mean()
salary_mean
df_filled = dr_drop.copy()
df_filled['Salary'].fillna(salary_mean, inplace=True)
df_filled
```

```python
df_with_dup = df_filled.copy()
df_with_dup = pd.concat([df_with_dup, df_with_dup.iloc[[0]]], ignore_index=True)
print(df_with_dup)
df_no_dup = df_with_dup.drop_duplicates()
print(df_no_dup)
```
